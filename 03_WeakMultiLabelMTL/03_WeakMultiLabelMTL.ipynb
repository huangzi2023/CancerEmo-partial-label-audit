# Weak Multi-Label Multi-Task Learning

Train one shared encoder with eight sigmoid heads on the outer-merged Dataset B under plain (weak-zero) `BCEWithLogitsLoss`. All non-positive cells in Dataset B are treated as negatives during training — this is the naive multi-task baseline that step 04's masked variant is compared against.

Wraps `scripts/mtl_weak_zero.py`. A T4 runs the full sweep in roughly 15-20 minutes at the default hyperparameters; A100 in under 10 minutes. **GPU required.**
